#### 1) Checking the features by double decompression

```python
import gzip

file = "Dataset/day-r5.2.csv.gz"

with gzip.open(file, "rb") as f1:
    with gzip.open(f1, "rt", errors="ignore") as f2:
        for i in range(5):
            print(f2.readline())
```

#### 3) Performing double decompression and saving as .csv file

```python
import gzip
import shutil

gz_file = "Dataset/day-r5.2.csv.gz"
csv_file = "Dataset/day-r5.2_raw.csv"

# Open First gzip layer
with gzip.open(gz_file, "rb") as f1:

    # Open Second gzip layer
    with gzip.open(f1, "rb") as f2:

        # Write actual CSV
        with open(csv_file, "wb") as out:
            shutil.copyfileobj(f2, out)

print("DOUBLE-GZIP extracted CSV saved!")
```
